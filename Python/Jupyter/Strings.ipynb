#Convertir a mayúsculas y minúsculas

En datos es común necesitar estandarizar los strings convirtiéndolos a minúsculas o mayúsculas para que Londres, londres y LONDRES sean considerados el mismo valor.

Esto también es útil cuando usamos condicionales basados en texto que ingresa un usuario.

Se usa:
- lower()
- upper()

```python
print(palabra.upper())
MayuscAMinusc = "HOLAAAA"
print(MayuscAMinusc.lower())
```

###Replace

Se pueden remplazar valores de strings

```python
skill = "python"
print(skill.replace("p","j"))
print(skill)

skill = "iiiiipii"
print("\n",skill.replace("i","x"))
print(skill)
```

Se pueden remplazar un número arbitrario de letras

```python
S = "Dntn Analist"
S2= S.replace("n","a", 2)
print(S2)
```

###Split

Muchas veces queremos separar strings

```python
print(S2)
S2.split(sep=" ", maxsplit=1)
```

```python
print(S2)
S2.split(sep="X", maxsplit=1)
```

```python
S2.split(sep="a", maxsplit=1)
```

```python
# El default maxsplit es -1 que significa cuantos splits sean necesarios
S2.split(sep="a", maxsplit=-1)
```

```python
W1 = "   ESPACIOS     "
W1.strip()
```

###Dunders

son métodos que empiezan y terminan con double underscore (__)

Algunos de estos se pueden escribir de forma más sencilla

```python
str.__add__("Data ","Analyst")
```

```python
"Data ".__add__("Analyst")
```

```python
"Data " + "Analyst"
```

```python
"Data " * 10
```

##Concatenación

Concatenación es combinar más de un solo string. Para ello se puede utilizar el `__add__` o el + que vimos antes

##Formatear

Existe el método format()
Aunque su función es muy similar a un fstring

```python
help(format)
```

```python
#Create the string "Role: Data Analyst"

role = "Data Analyst"
skill = "Python"

"Role: {}".format(role)
```

```python
# Creeate the string "Role: Data Analyst; Skill Required: Python"
"Role: {}; Skill Required: {}".format(role, skill)
```

```python
# Creeate the string "Role: Data Analyst; Skill Required: Python"
"Role: {role}; Skill Required: {skill}".format(role=role, skill=skill)
```

### fstring

Formatted string literals

```python
role = "Data Analyst"
skill = "Python"

f"Role: {role}; Skill Required : {skill}"
```

### %-formatting

Existe otra forma de imprimir, aunque no es tan usada, vale la pena conocerla

```python
role = "Data Analyst"
skill = "Python"

"Role: %s; Skill Required: %s" % (role, skill)
```

###Join

Concatena varios strings (funciona con cualquier iterable [un string es un iterable]).



```python
years_experience = "0123456789"
", ".join(years_experience)
```

```python
skills = ["Python","SQL","Excel"]

", ".join(skills)
```
