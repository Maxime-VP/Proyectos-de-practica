#Convertir a mayúsculas y minúsculas

En datos es común necesitar estandarizar los strings convirtiéndolos a minúsculas o mayúsculas para que Londres, londres y LONDRES sean considerados el mismo valor.

Esto también es útil cuando usamos condicionales basados en texto que ingresa un usuario.

Se usa:
- lower()
- upper()

```python
print(palabra.upper())
MayuscAMinusc = "HOLAAAA"
print(MayuscAMinusc.lower())
```

###Replace

Se pueden remplazar valores de strings

```python
skill = "python"
print(skill.replace("p","j"))
print(skill)

skill = "iiiiipii"
print("\n",skill.replace("i","x"))
print(skill)
```

Se pueden remplazar un número arbitrario de letras

```python
S = "Dntn Analist"
S2= S.replace("n","a", 2)
print(S2)
```

###Split

Muchas veces queremos separar strings

```python
print(S2)
S2.split(sep=" ", maxsplit=1)
```

```python
print(S2)
S2.split(sep="X", maxsplit=1)
```

```python
S2.split(sep="a", maxsplit=1)
```

```python
# El default maxsplit es -1 que significa cuantos splits sean necesarios
S2.split(sep="a", maxsplit=-1)
```

```python
W1 = "   ESPACIOS     "
W1.strip()
```

###Dunders

son métodos que empiezan y terminan con double underscore (__)

Algunos de estos se pueden escribir de forma más sencilla

```python
str.__add__("Data ","Analyst")
```

```python
"Data ".__add__("Analyst")
```

```python
"Data " + "Analyst"
```

```python
"Data " * 10
```
