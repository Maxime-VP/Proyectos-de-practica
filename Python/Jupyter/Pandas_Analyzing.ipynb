#groupby

```python
df.groupby("job_title_short")["salary_year_avg"].median()
```

Se pueden agrupar varias cosas

```python
df.groupby(["job_title_short", "job_country"])["salary_year_avg"].median()
```

```python
df.groupby("job_title_short")[["salary_year_avg", "salary_hour_avg"]].median()
```

También se pueden pedir varios valores agregados

```python
df.groupby("job_title_short")["salary_year_avg"].agg(["min","max","median","mean"])
```
