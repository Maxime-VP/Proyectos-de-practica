#Operadores aritméticos

Honestamente todos funcionan como uno lo esperaría creo que los únicos 3 que merecen ser explicados son el módulo %, la división entera // y la porencia **

- % da el residuo de una división
- // da el resultado entero de una division truncando siempre hacia abajo (no se redondea hacia arriba nunca)
- ** sirve para elevar a un número a un exponente

```python
5%2
```

```python
10 // 1.2
```

```python
2**5
```

## Assignment operators

Los operadores de asignación sirven para asignar un nuevo valor

`=, +=, -=, *=, /=, %=, **=, //=`

Funcionan básicamente igual que los aritméticos pero actualizan el valor original

```python
val = 1
val += 1
print(val)
```

```python
val = 2
val -= 2
print(val)
```

```python
val = 10
val **= 0
print(val)
```

```python
val = 10
val %= 2
print(val)
```

Se puede asignar más de un valor en una sola línea

```python
val1, val2 = 2, 3

f"{val1}, {val2}"
```

# Comparison Operators

Los operadores de comparación regresan True o False.

Como su nombre lo dice sirven para comparar dos valores.

`==, !=, >, <, >=, <=`

!= Compara si son diferentes, cuando son diferentes regresa True


```python
val1, val2 = 3,2

val1 == val2
```

```python
val1 != val2
```

```python
val1 <= val2
```

```python
val1 >= val2
```
