## Assignment operators

Los operadores de asignación sirven para asignar un nuevo valor

`=, +=, -=, *=, /=, %=, **=, //=`

Funcionan básicamente igual que los aritméticos pero actualizan el valor original

```python
val = 1
val += 1
print(val)
```

```python
val = 2
val -= 2
print(val)
```

```python
val = 10
val **= 0
print(val)
```

```python
val = 10
val %= 2
print(val)
```

Se puede asignar más de un valor en una sola línea

```python
val1, val2 = 2, 3

f"{val1}, {val2}"
```

#Operadores Lógicos

- and (ambas cosas deben cumplirse para ser true)
- or (con que al menos una se cumpla es true)
- not (si es true se vuelve false y viceversa)

```python
job_work_from_Home = True
job_health_insurance = True

print(job_work_from_Home and job_health_insurance)
print(job_work_from_Home or job_health_insurance)
print(not job_work_from_Home)
```

```python
job_work_from_Home = True
job_health_insurance = False

print(job_work_from_Home and job_health_insurance)
print(job_work_from_Home or job_health_insurance)
print(not job_work_from_Home)
```

```python
job_work_from_Home = False
job_health_insurance = False

print(job_work_from_Home and job_health_insurance)
print(job_work_from_Home or job_health_insurance)
print(not job_work_from_Home)
```

#Identity operators

Ayudan a ver si algun item está correlacionado con otro item, básicamente checan si tienen el mismo id

- is
- is not

```python
salary1 = 10000
salary2 = 10000

print(salary1 == salary2)
print(salary1 is salary2)
print(salary1 is not salary2)
print(salary1 is salary1)
```

```python
skills= ["Python", "SQL", "PowerBI"]
my_skills = skills

print(id(skills))
print(id(my_skills))
my_skills is skills
```
