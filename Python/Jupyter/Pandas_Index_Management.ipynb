# Index Management



```python
from datasets import load_dataset
import pandas as pd

ds = load_dataset("lukebarousse/data_jobs")
df = ds["train"].to_pandas()

df["job_posted_date"]= pd.to_datetime(df["job_posted_date"])

print(df.info())

df.head(1)
```

```python
df.sample(5)
```

```python
df.index
```

```python
df.index.name = "job_index"
```

```python
df.head(5)
```

# Vamos a revisar 3 métodos

- df.reset_index()
- df.set_index()
- df.sort_index()

## Reset index

```python
df_usa = df[df["job_country"]== "United States"]
df_usa.head()
```

```python
df_usa.reset_index(inplace=True)

df_usa.head(5)
```

Como se puede ver no se elimina el index original. En algunos casos es conveniente conservarlo para poder volver a unir la información con el dataset original.

# Set index

Asigna la columna seleccionada como el indice

```python
df_usa.set_index("job_index", inplace=True)
df_usa.head()
```

# Sort Index

Ordena el set por el indice

```python
median_pivot = df_usa.pivot_table(index="job_title_short", values="salary_year_avg", aggfunc=["median","min", "max"])
median_pivot
```

```python
median_pivot.sort_values(by=("median","salary_year_avg"), ascending=True, inplace=True)
median_pivot
```

```python
median_pivot.plot(kind="bar")
```

```python
median_pivot.sort_index(inplace=True)
```

```python
median_pivot.plot(kind="bar")
```
