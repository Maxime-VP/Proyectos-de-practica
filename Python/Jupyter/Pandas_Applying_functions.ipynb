# Apply

Apply permite aplicarle una función a una columna de nuestro dataset

```python
df = df.dropna(subset=["salary_year_avg"])
```

```python
df.head()
```

```python
def calculate_salary_inflation(salary, inflation = .05):
  return salary*(1+inflation)

sal = 10000

calculate_salary_inflation(sal)
```

```python
df["projected_salary_with_inflation"] = df["salary_year_avg"].apply(calculate_salary_inflation)
```

```python
df.sample(10)
```

## Convertir datos: [ast, (abstract sintax trees)]

```python
import ast
str_val = "['list']"
print(type(str_val))
list_val = ast.literal_eval(str_val)
print(type(list_val))
```

```python
def literal_evaluation(text):
  # alternativa:
  # if pd.notna(text) and type(text) == str:
  if text and type(text) == str:
    return ast.literal_eval(text)
  else:
    return text

print(type(df["job_skills"].iloc[0]))

df["job_skills"] = df["job_skills"].apply(literal_evaluation)

print(type(df["job_skills"].iloc[0]))
```

```python
df["job_skills"].info()
```

```python
#Cuidado con esto, las siguientes dos líneas no hacen lo mismo, iloc usa
#posición y [] usa el indice que puede perderse por hacer un dropna,
#por eso no son iguales al usar el mismo número y en algunos casos puede dar error
print(df["job_skills"].iloc[100])
df['job_skills'][100]
```

```python
df["job_skills"] = df["job_skills"].apply(
    lambda job_skills:
    ast.literal_eval(job_skills) if (job_skills and type(job_skills) == str)
    else job_skills
    )
```

# Filas
Actualmente apply lo hemos utilizado para modificar columnas, sin embargo se puede utilizar para filas también.

```python
def proyected_salary(row):
  if "senior" in row["job_title_short"]:
    return 1.05 * row["salary_year_avg"]
  return 1.03 * row["salary_year_avg"]
```

```python
df["salary_year_inflated"] = df.apply(proyected_salary, axis = 1)

df[["job_title_short","salary_year_avg","salary_year_inflated"]].sample(20, random_state=42)
```
