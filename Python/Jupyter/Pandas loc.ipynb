#Pandas loc

Pandas loc sirve para buscar valores dentro de nuestros datasets con cierto 

```python
import pandas as pd
from datasets import load_dataset
ds = load_dataset("lukebarousse/data_jobs")
```

```python
df = ds['train'].to_pandas()
df.head(3)
```

```python
df.info()
```

```python
df["job_posted_date"] = pd.to_datetime(df["job_posted_date"])
df.info()
```

```python
df = df.sort_values(by="job_posted_date", ascending=True)
df.head(3)
```

```python
df = df.dropna(subset=["salary_year_avg"])
```

```python
df.loc[:,"salary_rate":"salary_hour_avg"]

```

```python
df.loc[df["job_posted_date"].dt.strftime("%Y-%m-%d") == "2023-01-01"]
```
