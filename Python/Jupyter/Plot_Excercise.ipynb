##Vamos a hacer una gráfica para mostrar la mediana y la media de salario de cada trabajo

```python
import pandas as pd
from datasets import load_dataset
ds = load_dataset("lukebarousse/data_jobs")
```

```python
df = ds["train"].to_pandas()
df.head(1)
```

```python
df["job_posted_date"]=pd.to_datetime(df.job_posted_date)

df.info()
```

```python
df = df.sort_values("job_posted_date")
df.head()
```

```python
media_salario = df.groupby("job_title_short")["salary_year_avg"].mean().sort_values()
```

```python
mediana_salario =df.groupby("job_title_short")["salary_year_avg"].median().sort_values()
```

```python
media_salario.plot(kind="bar")
```

```python
mediana_salario.plot(kind="bar")
```
