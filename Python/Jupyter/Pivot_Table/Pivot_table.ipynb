# Pivot table

El pivot table de pandas recive 4 parámetros

- Values
- index
- columns
- aggfunc

```python
from datasets import load_dataset
import pandas as pd

ds = load_dataset("lukebarousse/data_jobs")
df = pd.DataFrame(ds["train"])

df.head(1)
```

```python
df["job_posted_date"] = pd.to_datetime(df["job_posted_date"])
df.info()
```

```python
df.pivot_table(index="job_title_short", aggfunc="size")
```

```python
df.pivot_table(index="job_title_short", values="salary_year_avg", aggfunc="median")
```

```python
df.pivot_table(index="job_title_short",values="salary_year_avg",columns="job_country", aggfunc="median")
```

La tabla anterior puede invertirse

```python
df.pivot_table(index="job_country",values="salary_year_avg",columns="job_title_short", aggfunc="median")
```

```python

top_countries = df["job_country"].value_counts().head(5).index
top_countries
```

```python
df_job_country_salary = df.pivot_table(values="salary_year_avg", index="job_country", columns="job_title_short", aggfunc="median")

df_top_job_countries_salary = df_job_country_salary.loc[top_countries]

df_top_job_countries_salary
```

```python
job_titles = ["Data Analyst", "Data Engineer", "Data Scientist"]
df_top_job_countries_salary_top_titles = df_top_job_countries_salary[job_titles]
df_top_job_countries_salary_top_titles
```

```python
df_top_job_countries_salary_top_titles.plot(kind="bar", title="Median salary by job title and country", figsize=(10, 6))
```
