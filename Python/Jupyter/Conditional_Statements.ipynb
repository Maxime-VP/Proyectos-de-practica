#Conditional statements

if
elif
else

Estos nos permiten hacer validaciones.

if nos permite comprobar si una condición es verdadera o falsa, si es verdadera ejecutará el código dentro.

elif sirve para cuando quieres comparar varios casos excluyentes, el primer caso sera un if y el segundo y los posteriores serán elif. Cuando se cumple uno de los if o elif en el bloque el resto no se evaluarán.

Es posible emplear varios if uno seguido de otro, pero se evaluan cada uno de forma independiente. Si son casos excluyentes es mejor utilizar elif, pero dependerá del contexto y el comportamiento deseado del código.

El else sirve cuando queremos abarcar todos los casos que no están dentro de nuestros if y elif, este es útil para mensajes de error o cuando el código tiene un comportamiento predecible y determinado para las condiciones restantes.

```python
job= "Data Engineer"
if job == "Data Engineer":
  print("Congratulations")
```

```python
years_of_experience=4
if years_of_experience <=1:
  print("jr")
elif years_of_experience <=3:
  print("intermediate")
else:
  print("sr")
```
