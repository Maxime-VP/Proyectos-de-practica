# df.sample() y df.copy()

```python
import pandas as pd
from datasets import load_dataset
ds = load_dataset("lukebarousse/data_jobs")
df = ds["train"].to_pandas()
df["job_posted_date"] = pd.to_datetime(df["job_posted_date"])
df = df.sort_values(by="job_posted_date", ascending=True)
df.head()
```

#df.sample()

Regresa una muestra de items de un eje (axis) de un objeto.

```python
df.sample() # Regresa 1 elemento
```

```python
df.sample(10)
```

El valor que regresa es aleatorio pero podemos agregar una semilla con el parámetro random_state

```python
df.sample(random_state=42)
```

```python
df.sample(random_state=42)
```

# df.copy()

Crea una copia de los indices y data de un objeto. Si no usamos copy(), al hacer un nuevo df, si queremos actualizar nuestro nuevo df se alterará el original también.

Para comprobar esto descomenta y comenta las líneas donde se encuentra el copy

```python
ds = load_dataset("lukebarousse/data_jobs")
df_original = ds["train"].to_pandas()
```

```python
#Cambia la línea no comentada por la comentada:
# ----------------------------------------
df_altered = df_original.copy()
#df_altered = df_original
# ----------------------------------------

df_altered
```

```python
df_altered.loc[:5,"salary_year_avg"]
```

```python
median_salary = df_altered["salary_year_avg"].median()
median_salary
```

```python
df_altered["salary_year_avg"] = df_altered["salary_year_avg"].fillna(median_salary)

df_altered.loc[:5,"salary_year_avg"]
```
