#Pandas Plot
Pandas tiene funciones para crear gráficas

```python
import pandas as pd
from datasets import load_dataset
ds = load_dataset("lukebarousse/data_jobs")

df = ds["train"].to_pandas()

df.head(1)
```

```python
df["job_posted_date"]= pd.to_datetime(df.job_posted_date)
df.info()
```

```python
df["job_posted_month"]= df.job_posted_date.dt.month
df.head()
```

```python
job_count= df.job_title_short.value_counts()
job_count = job_count.sort_values(ascending=False)
job_count
```

```python
job_by_month_count = df["job_posted_month"].value_counts()
job_by_month_count = job_by_month_count.sort_index()
job_by_month_count
```

```python
job_by_month_count.plot()
```

```python
job_count.plot(kind="bar")
```

##Matplotlib
Podemos emplear matplotlib sobre el plot de Pandas

```python
import matplotlib.pyplot as plt
job_count.plot(kind="bar")
plt.title("Job posting count by title")
plt.xlabel("")
plt.ylabel("Count")
plt.xticks(rotation=45, ha="right")
plt.show()
```

```python
df = df.sort_values(by="job_posted_date")
df.head()
```

```python
df[["job_posted_date", "salary_year_avg"]].dropna(subset=["salary_year_avg"])
```

```python
df.plot(x="job_posted_date",y="salary_year_avg", kind= "line")
```

#Voy a agrupar por mes para graficar mejor

```python
df["posted_date_year_month"] = df.job_posted_date.dt.to_period("M")
```

```python
mean_salary_per_month = (
    df.groupby("posted_date_year_month")
      ["salary_year_avg"].mean()
)
mean_salary_per_month
```

```python
mean_salary_per_month.plot(kind="line")
```
