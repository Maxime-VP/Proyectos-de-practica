# Clase salario

Este notebook es un ejemplo de una clase de python, con métodos dunder init y repr.

`:,.0f` sirve para decir cuántos decimales queremos

`__repr__` sirve para dar una descripción del objeto, en este caso simplemente devuelve el salario.

```python
class Salary:
  def __init__(self, base_salary, bonus_rate = .1, symbol = "$"):
    self.base_salary = base_salary
    self.bonus_rate = bonus_rate
    self.symbol = symbol

  def __repr__(self):
    return f"{self.symbol}{self.base_salary:,.0f}"

  def total_pay(self):
    return f"{self.symbol}{(self.base_salary*(1+self.bonus_rate)):,.0f}"

  def total_bonus(self):
    return f"{self.symbol}{self.base_salary*self.bonus_rate:,.0f}"
```

```python
mi_salario = Salary(20000, .2)
mi_salario.symbol
```

```python
#Ejecuta el __repr__
mi_salario
```

```python
mi_salario.total_pay()
```

```python
mi_salario.total_bonus()
```
