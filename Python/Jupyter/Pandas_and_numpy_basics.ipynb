#Numpy

numpy es una libreria de python para computación científica y manejo eficiente de datos numéricos mediante arrays multidimensionales.

##Ventajas grandes de numpy
- es mucho más ráoido que otras librerias
- es compatible con muchas bibliotecas
  - Visualización: seaborn, matplotlib
  - Ciencia de datos: pandas, scikit-learn

```python
import random
salary_list = [random.randint(50000,100000) for i in range(1000000)]
```

##Usaremos la librería de statistics y %%timeit para comparar la ejecución de ambas librerias

```python
import statistics
```

```python
%%timeit

statistics.mean(salary_list)
```

```python
import numpy as np
```

```python
%%timeit

np.mean(salary_list)
```

Numpy es mucho más rápido para ejecutar el mismo cálculo, a pesar de ser milisegundos, cuando se trabaja con miles de millones de datos un trabajo

```python
job_titles = np.array(["Software dev","Data eng","Data Analyst","Data Scientist","ML engineer"])

base_salaries =  np.array([20000, 30000, 10000, 45000, 40000])

bonus_rates =  np.array([.14, .12, .11, .15, .16])
```

Podemos usar nan para valores que no tenemos. Se llama con np.nan

```python
job_titles = np.array(["Software dev","Data eng","Data Analyst","Data Scientist","ML engineer", "AI engineer"])

base_salaries =  np.array([20000, 30000, 10000, 45000, 40000, np.nan])

bonus_rates =  np.array([.14, .12, .11, .15, .16, np.nan])
```

```python
total_salaries = base_salaries * (1 + bonus_rates)

total_salaries
```

```python
# El metodo mean no funciona con valores nan, para ello debemos utilizar nanmean
#np.mean(total_salaries)

np.nanmean(total_salaries)

#El resultado debe ser igual al anterior ya que no se cuentan los valores nan en el proceso
```

# Pandas

Pandas sirve para trabajar con datos de manera eficiente y flexible en Python.

- Limpieza y tratamiento de datos: Ideal para preparar datos eliminando valores nulos, duplicados o inconsistencias.

- Análisis exploratorio de datos (EDA): Facilita la visualización y el análisis inicial de los datos para identificar patrones o tendencias.

- Soporte para Machine Learning: Sirve como base para preparar datos antes de entrenar modelos.

- Consultas en bases de datos: Permite realizar consultas y transformaciones en datos relacionales.

- Visualización de datos: Se integra con bibliotecas como Matplotlib y Seaborn para crear gráficos.

- Lectura y escritura de archivos: Compatible con formatos como CSV, Excel, JSON, SQL, entre otros.
