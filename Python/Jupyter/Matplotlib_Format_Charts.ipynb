Subplots permite poner varias gráficas en un solo espacio:

```python
import pandas as pd
import matplotlib.pyplot as plt
from datasets import load_dataset

ds = load_dataset("lukebarousse/data_jobs")
```

```python
df = ds["train"].to_pandas()
df.head(1)
```

```python
import ast

def literal_eval(cadena):
  if cadena and type(cadena) == str:
    return ast.literal_eval(cadena)
  return cadena

df["job_posted_date"] = pd.to_datetime(df["job_posted_date"])
df["job_skills"] = df["job_skills"].apply(literal_eval)
df.info()
```

```python
type(df.job_skills[100])
```

```python
df.head()
```

```python
df.sort_values(by="job_posted_date", inplace = True)
```

```python
df.head()
```

# Matplotlib

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(2,3)
```

```python
fig, ax = plt.subplots()

df["job_title_short"].value_counts().plot(kind="bar", ax = ax)
```

# Pandas con plt

```python
fig, ax = plt.subplots(1,2)

df["job_title_short"].value_counts().plot(kind="bar", ax= ax[0])
df["job_schedule_type"].value_counts().head(3).plot(kind="bar", ax = ax[1])

fig.tight_layout()
```

```python
fig, ax = plt.subplots(1,2)

ax[0].bar(df["job_title_short"].value_counts().index, df["job_title_short"].value_counts())
ax[1].bar(df["job_schedule_type"].value_counts().head(3).index, df["job_schedule_type"].value_counts().head(3))

fig.tight_layout()
```

# Top skills per job

```python
top_jobs = df["job_title_short"].value_counts().head(5).index.to_list()
```

```python
top_jobs
```

```python
df_top_jobs = df[df["job_title_short"].isin(top_jobs)]
df_top_jobs.sample(20)
```

```python
df_top_skills_exploded = df_top_jobs.explode("job_skills")
df_top_skills_exploded.head()
```

```python
top_skills = df_top_skills_exploded["job_skills"].value_counts().head(5).index.to_list()
```

```python
top_skills
```

```python
df_reduced = df_top_skills_exploded[df_top_skills_exploded["job_skills"].isin(top_skills)]
df_reduced.head()
```

```python
df_data_analyst = df_reduced[df_reduced["job_title_short"] == "Data Analyst"]
df_data_analyst.head()
```

```python
df_Data_Engineer = df_reduced[df_reduced["job_title_short"] == "Data Engineer"]
df_Data_Engineer.head()
```

```python
df_Data_Scientist = df_reduced[df_reduced["job_title_short"] == "Data Scientist"]
df_Data_Scientist.head()
```

```python
# En lugar de sharey, se puede usar ylim o xlim si está invertida la tabla para definir un mínimo y máximo común
#ylim=(0,120000)

fig, ax = plt.subplots(1,3, sharey=True)
df_data_analyst["job_skills"].value_counts().plot(kind="bar", ax = ax[0], xlabel="Data Analyst")
df_Data_Engineer["job_skills"].value_counts().plot(kind="bar", ax = ax[1], xlabel="Data Engineer")
df_Data_Scientist["job_skills"].value_counts().plot(kind="bar", ax = ax[2], xlabel="Data Scientist")


fig.suptitle("Top skills per job")
fig.tight_layout()
```

## Como acceder a otros niveles

```python
fig, ax = plt.subplots(2,2, sharey=True)
df_data_analyst["job_skills"].value_counts().plot(kind="bar", ax = ax[0,0], xlabel="Data Analyst")
df_Data_Engineer["job_skills"].value_counts().plot(kind="bar", ax = ax[0,1], xlabel="Data Engineer")
df_Data_Scientist["job_skills"].value_counts().plot(kind="bar", ax = ax[1,0], xlabel="Data Scientist")

fig.suptitle("Top skills per job")
fig.tight_layout()
```

```python
# En lugar de sharey, se puede usar ylim o xlim si está invertida la tabla para definir un mínimo y máximo común
#ylim=(0,120000)

fig, ax = plt.subplots(1,3)
df_data_analyst["job_skills"].value_counts().plot(kind="bar", ax = ax[0], xlabel="Data Analyst", ylim=(0,120000))
df_Data_Engineer["job_skills"].value_counts().plot(kind="bar", ax = ax[1], xlabel="Data Engineer", ylim=(0,120000))
df_Data_Scientist["job_skills"].value_counts().plot(kind="bar", ax = ax[2], xlabel="Data Scientist", ylim=(0,120000))

ax[0].invert_xaxis()
ax[1].invert_xaxis()
ax[2].invert_xaxis()

fig.suptitle("Top skills per job")
fig.set_figwidth(15)
fig.tight_layout()
```

```python
# En lugar de sharey, se puede usar ylim o xlim si está invertida la tabla para definir un mínimo y máximo común
#ylim=(0,120000)

fig, ax = plt.subplots(1,3)
df_data_analyst["job_skills"].value_counts().plot(kind="barh", ax = ax[0], xlabel="Data Analyst", xlim=(0,120000))
df_Data_Engineer["job_skills"].value_counts().plot(kind="barh", ax = ax[1], xlabel="Data Engineer", xlim=(0,120000))
df_Data_Scientist["job_skills"].value_counts().plot(kind="barh", ax = ax[2], xlabel="Data Scientist", xlim=(0,120000))

fig.suptitle("Top skills per job")
fig.set_figwidth(15)
fig.tight_layout()
```

```python
fig, ax = plt.subplots(1,3)

df_data_analyst["job_skills"].value_counts().sort_index().plot(
    kind="bar", ax=ax[0], xlabel="Data Analyst", xlim=(0, 120000)
)

df_Data_Engineer["job_skills"].value_counts().sort_index().plot(
    kind="bar", ax=ax[1], xlabel="Data Engineer", xlim=(0, 120000)
)

df_Data_Scientist["job_skills"].value_counts().sort_index().plot(
    kind="bar", ax=ax[2], xlabel="Data Scientist", xlim=(0, 120000)
)

fig.suptitle("Top skills per job")
fig.set_figwidth(12)
fig.tight_layout()
```
