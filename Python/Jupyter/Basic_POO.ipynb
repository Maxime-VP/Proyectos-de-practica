## Definicion de una clase

Dunder = double underscore (__)

Los dunder son métodos especiales que empiezan y terminan con __

`__init__` se ejecuta cuando se crea un objeto



```python
class JobPost:
    def __init__(self, title, location, salary):
      self.title = title
      self.location = location
      self.salary = salary
```

```python
job_title_ex = "Data Analyst"
job_location_ex = "Mexico"
job_salary_ex = 30000

JobPost(job_title_ex, job_location_ex, job_salary_ex)
```

###Atributos

Un campo perteneciente a una clase

```python
class JobPost:
    def __init__(self, title, location, salary):
      #Atributos:
      self.title = title
      self.location = location
      self.salary = salary
```

```python
job1 = JobPost(job_title_ex, job_location_ex, job_salary_ex)

print(job1.title)
print(job1.salary)
```

### Metodos

Un método es una función definida dentro de una clase que opera con la información en sus objetos.

```python
class JobPost:
    def __init__(self, title, location, salary):
      #Atributos:
      self.title = title
      self.location = location
      self.salary = salary
      #Métodos
      self.print_data = self.print_data

    def print_data(self):
      print("Titulo    ", self.title)
      print("Ubicacion ", self.location)
      print("Salario   ", self.salary)
```

```python
job1 = JobPost("Data Analyst", "Mexico", "30000")

job1.print_data()
```

```python
help(job1)
```
