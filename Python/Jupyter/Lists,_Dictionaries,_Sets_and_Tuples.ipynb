# Lists, Dictionaries, Sets and Tuples

Son formas distintas de almacenar datos en python

Las listas son las más básicas simplemente guardan los elementos.

Los diccionarios son pares de valores, una llave y un valor relacionado a esa llave.

Los sets son prácticamente listas que no permiten valores duplicados.

Las tuplas son parecidas a las listas con la particularidad de que no se pueden modificar.



##Listas

Las listas permiten agregar y remover elementos. Además nos permiten acceder a los elementos por su índice (comienza en 0).

- append() agrega un valor
- remove() elimina la primera instancia del elemento seleccionado
- len(lista) nos da la longitud de la lista


```python
lista = ["uno", "dos","cuatro"]
```

```python
lista.append("tres")
print(lista)
```

```python
lista.remove("cuatro")
print(lista)
```

```python
len(lista)
```

```python
lista[1]
```

###Slicing

syntax: list[start:end:step]

start es el indice donde inicia
end es el indice donde termina
step es el número de pasos entre cada item


```python
print(lista)
print(lista[0:3])
```

```python
print(lista[1:1])
print(lista[1:2])
```

```python
print(lista[0:3:2])
```

```python
numbers1 =[1,2,3,4,5]
numbers2 =[4,5,6,7,8]
numbers3= numbers1 + numbers2
# Se repiten valores ya que en las listas es válido
print(numbers3)

#Se pueden acceder a valores usando un índice negativo
print(numbers3[-1])
# Todos menos el último
print(numbers3[:-1])
```

```python
for number in numbers3:
  print(number)
```

###Unpacking

Se puede asignar un valor nuevo a partir de una lista

```python
fruits=["apple","banana","lime"]
fruit1, fruit2, fruit3 = fruits
print(fruit1)
print(fruit2)
print(fruit3)
```

Se pueden empacar en una lista usando el operador *

```python
fruits=["apple","banana","lime"]
bestfruit, *notgreatfruits, = fruits
print(bestfruit)
print(notgreatfruits)
#The best fruit is mango btw, its just an example
```

#Dictionaries {}

Los diccionarios permiten alamacenar valores asociándolos con una llave.

Los valores pueden repetirse, pero las llaves no.

```python
dictionary = {
    1:"one",
    2:"two",
    3:"three"
}

print(dictionary[2])
```

```python
job_type_skills = {
    "Databases":"SQL",
    "Cloud":"AWS",
    "Libraries":"three"
}

job_type_skills["Cloud"]
```

Podemos usar una lista como datos, pero no como llave

```python
dict2 = {
    "number":5,
    "list":[1,2,3,4]
}

print(dict2["number"])
print(dict2["list"])
```

```python
print(dictionary)
```

```python
help(dict)
```

```python
dictionary.keys()
```

```python
dictionary.items()
```

```python
dict3 = {
    "uno":1,
    "dos":2,
    "tres":3
}

#Pop saca el valor de la llave
print(dict3.pop("dos"))
print(dict3)
```

```python
dict4 = {
    "uno":1,
    "dos":2,
    "tres":3
}

dict4.update({"cuatro":4})
dict4.update({"dos":"two"})
print(dict4)
```

Otra forma de agregar valores es así

```python
dict4["cinco"] = 5
dict4["dos"] = 2
print(dict4)
```

#Sets {}

Son similar a las listas, pero se eliminan los valores repetidos. Comparten las {} con los diccionarios, pero no tienen los dos puntos que marcan la relación ":"

No tienen indices

```python
skills_list = {"SQL","SQL","Python","Python","Python"}
print(skills_list)
```

```python
skills_list.add("SQL")
skills_list.add("Added")
print(skills_list)
```

```python
skills_list.remove("Added")
print(skills_list)
```

```python
#El pop elige un valor random a remover
skills_list.pop()
print(skills_list)
```

```python
lista_data = ["SQL","SQL", "SQL", "SQL", "SQL","SQL"]
print(lista_data)
set_data = set(lista_data)
print(set_data)
```

#Tuples ()

Son similares a las listas, pero inmutables, es decir no se pueden modificar.

Nota: la única forma de modificar una tupla es cuando contienen una lista, pero solo podemos cambiar el valor de dentro.

No se le pueden agregar ni eliminar valores, solo se pueden consultar

```python
help(tuple)
```

```python
skills_unmutable = ("Python", "SQL", "Statistics", "Tableau")
skills_unmutable
```

```python
skills_unmutable[1]
```

```python
skills_unmutable[0:2]
```

Aqui está la forma de cambiar el valor de una tupla

```python
set_withlist = ([1,2,3],[1,2,3])
set_withlist[1][2]="Changed"
print(set_withlist)
```

Puedes combinar tuplas



```python
tuple1 = (1,2,3)
tuple2 = (4,5,6)
tuple3 = tuple1 + tuple2

print(tuple3)
```

También se puede combinar sobre la tupla original, solo que esto genera una tupla nueva. Esto se puede ver por el id.

```python
tuple1 = (1,2,3)
print(id(tuple1))
tuple2 = (4,5,6)
tuple1 += tuple2

print(id(tuple1))
print(tuple1)
```
