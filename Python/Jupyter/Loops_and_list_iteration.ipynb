#Loops

Los ciclos nos ayudan a correr código n número de veces, esto es muy útil cuando necesitamos hacer alguna operación un número arbitrario de veces seguidas.

```python
jobs= ["Data Analyst", "Data engineer", "Software developer"]
```

```python
analyst_list = []
for job in jobs:
  if "Data Analyst" in job:
    analyst_list.append(job)
```

```python
print(analyst_list)
```

```python
numbers = []
for i in range(1,10+1):
  numbers.append(i)
print(numbers)
```

```python
years_experience = {
    "Luke": 3,
    "Mike": 6,
    "Jake": 4,
    "Zeke": 1,
    "Blake": 0
}
```

##Dos formas de interactuar con diccionarios

- .values()
- .keys()
- .items()

```python
for person in years_experience:
  print(years_experience[person])
```

```python
for person in years_experience.values():
  print(person)
```

```python
for key in years_experience.keys():
  print(key)
```

```python
for key, value in years_experience.items():
  print(key, value)
```

```python
for key, value in years_experience.items():
  if value > 3:
    print(key, value)
```

# While

```python
i=0
while True:
  print("Hello")

  i += 1
  if i>=5:
    break
```

```python
years_experience = {
    "Luke": 3,
    "Mike": 6,
    "Jake": 4,
    "Zeke": 1,
    "Blake": 0
}

years_list = list(years_experience.items())
print(years_list)
years = 0
index = 0
while years < 5 and index < len(years_list):
  key, years = years_list[index]
  print(key, years)
  index += 1
```

# List comprehension



```python
numbers = []
for i in range(100):
  numbers.append(i)
print(numbers)
```

Simplificado

```python
numbers= [x for x in range(100)]
print(numbers)
```

```python
numbers= [x*2 for x in range(50)]
print(numbers)
```

```python
letters = [x for x in "Python"]
print(letters)
```

Se puede poner ifs en la misma línea

```python
letters = [x for x in "Python" if "y" in x ]
print(letters)
```
