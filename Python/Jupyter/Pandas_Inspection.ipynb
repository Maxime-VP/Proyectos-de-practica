###unique()

unique nos sirve para identificar los diferentes valores de una columna. Con estos podemos detectar por ejemplo si una columna solo tiene 1 valor.

```python
job_titles = df_salario_limpio.job_title_short.unique()
print(job_titles)
```

```python
skills = df_salario_limpio.job_skills.unique()
print(skills)
#NOTA: en este ejemplo las listas contienen combinaciones de skills, por ello se consideran valores distintos y se repiten elementos como python
```

Se puede hacer una versión mejor de esto, estoy utilizando ast para corregir las listas que entraron como strings

```python
import ast

skills = df_salario_limpio.job_skills.unique()
skillset_set = set()

for skill_list in skills:
    if skill_list is not None:
        skill_list = ast.literal_eval(skill_list)

        for skill in skill_list:
            skillset_set.add(skill)

print(sorted(skillset_set))
```

```python
print(len(skillset_set))
```
