```python
import tensorflow as tf
import keras
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
mnist = tf.keras.datasets.mnist

(X_train, y_train),(X_test, y_test) = mnist.load_data()
```

# Normalizar los datos

```python
X_train = X_train / 255
X_test = X_test / 255
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout, Flatten, Input
from tensorflow.keras.layers import Conv2D, MaxPooling2D
```

```python
model = Sequential(
    [
        Input(shape=(28,28)),
        Flatten(),
        Dense(128, activation='relu'), # 128 son las neuronas, mientras más grande puede procesar cosas más complejas pero tarda más y puede causar overfitting
        # Relu pone todos los números menores a 0 en 0 y los números mayores a cero lo dejan así
        Dropout(0.2),
        Dense(10)
    ]
)
```

```python
nonsense_prediction = model(X_train[:1]).numpy()
nonsense_prediction
```

```python
tf.nn.softmax(nonsense_prediction).numpy().sum()
```

```python
from tensorflow.keras.losses import SparseCategoricalCrossentropy
loss_fn = SparseCategoricalCrossentropy(from_logits=True)

loss_fn(y_train[:1], nonsense_prediction).numpy()
```

```python
from tensorflow.keras.optimizers import Adam

optimizer = Adam(learning_rate=0.01)

model.compile(optimizer=optimizer,
              loss=loss_fn,
              metrics=['accuracy'])
```

```python
model.fit(X_train, y_train, epochs=10)
```

```python
model.evaluate(X_test,  y_test, verbose=2)
```

```python
tf.nn.softmax(model(X_test[:1])).numpy()
```
