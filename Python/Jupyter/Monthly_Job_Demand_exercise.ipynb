# Monthly job demand by month

```python
from datasets import load_dataset
import pandas as pd

ds = load_dataset("lukebarousse/data_jobs")
df = ds["train"].to_pandas()
df.head(1)
```

```python
df["job_posted_date"] = pd.to_datetime(df.job_posted_date)
df.info()
```

```python
df_sorted = df.sort_values(by="job_posted_date", ascending=True)
df_sorted.head()
```

```python
df_sorted["month"] = df_sorted["job_posted_date"].dt.month
```

```python
df_sorted.sample(12)
```

```python
df_pivot = df_sorted.pivot_table(index="month", values="job_title_short", aggfunc="count")
df_pivot
```

```python
df_pivot.plot(kind="bar")
```

```python
unique_jobs = df_sorted["job_title_short"].unique()
unique_jobs
```

```python
df_grouped = df_sorted.groupby(["month","job_title_short"]).size()
df_grouped
```

```python
df_grouped_jan = df_grouped.loc[1]
```

```python
df_grouped_jan.plot(kind="bar", title="January jobs", xlabel="")
```

```python
df_grouped_dic = df_grouped.loc[12]
```

```python
df_grouped_dic.plot(kind="bar", title="December jobs", xlabel="")
```

```python
df_grouped_Data_Analyst = df_grouped.loc[
    (slice(None), "Data Analyst")
]
```

```python
df_grouped_Data_Analyst.plot(kind="bar")
```

# Alternativa

Esta es un poco mejor, pero debemos de tener cuidado al ordenar los valores.

```python
df_sorted["month"] = df_sorted["job_posted_date"].dt.strftime("%B")
df_sorted.head()
```

```python
df_pivot = df_sorted.pivot_table(index="month", columns="job_title_short", aggfunc="size")
df_pivot.reset_index(inplace=True)
df_pivot
```

Reconstruir el índice

```python
df_pivot["month_num"] = pd.to_datetime(df_pivot["month"], format="%B").dt.month
df_pivot.sort_values(by="month_num", inplace=True)
df_pivot.set_index("month", inplace = True)
```

```python
df_pivot.drop(columns="month_num", inplace=True)
df_pivot
```

```python
df_pivot.plot(kind="bar", figsize=(20, 6))
```

```python
top_3 = df_sorted["job_title_short"].value_counts().head(3)
top_3 = top_3.index.to_list()
```

```python
df_pivot[top_3].plot(kind="bar")
```
