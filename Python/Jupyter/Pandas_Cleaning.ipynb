#Limpieza de datos

##Cambiar a datetime

En nuestro Dataset tenemos un datetime que está como un string, vamos a cambiarlo a datetime

```python
df.job_posted_date.info()
```

```python
df["job_posted_date"] = pd.to_datetime(df.job_posted_date)
```

```python
df.job_posted_date.info()
```

Al tener la columna correctamente como datetime (dt), tenemos acceso a funciones de fechas que nos pueden por ejemplo servir para extraer el mes

```python
df.job_posted_date.dt.month
```

```python
#El siguiente código no sirve para crear una nueva columna, para ello requerimos usar [""]
df.job_posted_month = df.job_posted_date.dt.month
```

```python
df.head(1)
```

```python
df["job_posted_month"] = df.job_posted_date.dt.month
```

```python
df.head(1)
```

###Sort dataframa:

pandas.DataFrame.sort_values

```python
df.sort_values("job_posted_date")
```

El código anterior no cambia el orden del df original. Para eso requerimos agregar el parametro inplace

```python
df.head()
```

```python
df.sort_values("job_posted_date", inplace=True)
```

```python
df.head()
```

En el código que sigue, nuevamente falta el inplace, pero es la forma en la que se elimina una columna.
