#Modulos

Python nos permite importar modulos para poder emplear datos o funciones preexistentes. Estos pueden ser tanto bibliotecas ya existentes como creadas por nosotros mismos.

Podemos darle un alias a las bibliotecas cuando son muy largas o se emplean mucho puede resultar más cómodo

```python
import modulesMax as mM

mM.suma(5.5,6)

#Con esta forma de importar lo siguiente no funcionaría
#suma(3,4)
```

Hay varias formas de hacer la importación, la siguiente forma sirve para importar funciones o variables en específico.

Esto nos sirve cuando solo requerimos algunas funciones, útil para

```python
from modulesMax import resta

resta(10,6)
```

```python
from modulesMax import lista

print(lista)
```

# Cuando una biblioteca no es pesada y la conocemos bien podemos importar todo con el *, alternativamente también podemos importar varias cosas en una línea con el uso de comas

```python
from modulesMax import lista, resta, suma

from modulesMax import *


#Con esta importación ya podemos emplear
suma(11,6)
```

La función help puede usarse para consultar cómo funciona un método importado pero para que funcione el módulo debe de contar con documentación.
Esto se logra con un comentario multilínea.

La función help igual puede utilizarse pero no regresará información si no está documentado.

En la documentación es común ver args y returns que nos sirven para ver los argumentos de entrada esperados con sus tipos de datos y el tipo de valor que retorna y qué es.

```python
help(suma)
```

Sin documentación:

```python
help(resta)
```

Los objetos importados también pueden ser consultados

```python
help(lista)
```

También podemos usar help() sobre todo el módulo:

```python
help(mM)
```

```python
import math
help(math)
```
