##Datatypes

Los data types (tipos de datos) representan las diferentes formas en las que podemos almacenar y trabajar con información.

Python es bastante amigable con los tipos de datos, ya que normalmente determina automáticamente el tipo de una variable según el valor que le asignamos.

También podemos especificar o convertir explícitamente un valor a cierto tipo.

- Para los números existen distintos tipos de datos que representan el tipo de número ya sean enteros (int), decimales (float) o complejos.
Usualmente solamente se utilizan los int y los floats.

- Otros tipos de variables numéricas como double o long int no existen en python 3.

- Los otros dos tipos de dato más relevantes son para texto (string) y para relaciones de verdadero y falso (bool). En los booleanos a veces se representan los valores verdaderos (True) con un 1 y los falsos (False) con un 0.

```python
roll_number = 1
height = 1.92
name = "Me"
is_student = True
```

```python
print(type(roll_number))
print(type(height))
print(type(name))
print(type(is_student))
```

Podemos especificar el tipo

```python
edad = int("25")
altura = float("1")
numero = str(100)

var_booleano = bool(1)
```

```python
print(type(edad))
print(type(altura))
print(type(numero))
print(numero)

print(type(var_booleano))
print(var_booleano)
```

##Comandos mágicos

Al escribir % nos aparecerá una lista de comandos.

- El comando %timeit nos permite ver cuánto tarda en ejecutar nuestro código.
  - Con un solo % funciona para una línea, con dos % (%%) sirve para toda la celda.

```python
%timeit [i for i in range(10)]
%timeit [i for i in range(1000)]
```

```python
%%timeit
for i in range(10):
  i
for i in range(1000):
  i
```
