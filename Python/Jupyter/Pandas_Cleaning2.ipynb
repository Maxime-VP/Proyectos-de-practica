# Data imputation

Imputar datos es rellenar los valores vacios con un aproximado que puede provenir de una fuente externa, de un promedio o de una media.

No siempre es buena idea imputar datos, pero en muchos casos es algo razonable (por ejemplo cuando la columna puede construirse a partir de informacíón que sí tenemos).

- fillna()

```python
import pandas as pd
from datasets import load_dataset
ds = load_dataset("lukebarousse/data_jobs")
df = ds["train"].to_pandas()
df["job_posted_date"] = pd.to_datetime(df["job_posted_date"])
df = df.sort_values(by="job_posted_date", ascending=True)
df.head()
```

Existen los métodos ffill y bfill que quieren decir forward fill y backwards fill respectivamente. Estos lo que hacen es rellenar el valor con el último valor válido (ffill) o con el siguiente valor válido (bfill).

```python
# df_filled = df.fillna(method="ffill")
```

Sin embargo muchas veces es mejor usar la mediana, un promedio o algúna otra estimación. En el caso del dataset que tenemos es mala idea asumir que los valores serán iguales a los anteriores o posteriores.

```python
median_year_salary = df["salary_year_avg"].median()
median_hourly_salary = df["salary_hour_avg"].median()

df_filled = df.copy()
df_filled["salary_year_avg"] = df_filled["salary_year_avg"].fillna(median_year_salary)
df_filled["salary_hour_avg"] = df_filled["salary_hour_avg"].fillna(median_hourly_salary)

df_filled.head()
```

# Dropping duplicates

A veces tenemos valores que son exactamente iguales y no vale la pena tenerlos dos veces


```python
df_unique = df_filled
print(df_filled.shape)
df_unique =df_unique.drop_duplicates()
print(df_unique.shape)

print("Valores eliminados: ", df_filled.shape[0] - df_unique.shape[0])
df_unique.head()
```

## También podemos ser más específicos en lo que consideramos un valor duplicado

```python
df_unique = df_filled
print(df_filled.shape)


df_unique = df_filled.drop_duplicates(subset=["job_title", "company_name"])


print(df_unique.shape)

print("Valores eliminados: ", df_filled.shape[0] - df_unique.shape[0])
df_unique.head()
```
