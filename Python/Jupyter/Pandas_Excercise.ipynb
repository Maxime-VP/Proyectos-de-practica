# Este es un ejercicio de groupby

```python
from datasets import load_dataset
import pandas as pd

ds = load_dataset("lukebarousse/data_jobs")
ds.keys()
```

```python
df = ds["train"].to_pandas()
df.head()
```

Filtrar únicamente en México

```python
# En el dataset todos son Mexico, pero en otro dataset podría no ser cierto
df_mexico = df[(df.job_country == "Mexico") | (df.job_country == "México" )]
df_mexico.head()
```

```python
df_mexico.shape
```

```python
from numpy import nan
df_mexico = df_mexico[df_mexico.salary_year_avg.notna()]
```

```python
df_mexico.shape
```

```python
df_mexico.job_title_short.value_counts()
```

```python
df_mexico.groupby("job_title_short").count()
```

```python
df_mexico.groupby(["job_title_short"])["salary_year_avg"].agg(["min","max", "mean", "median", "count"]).sort_values(by= "median")
```
