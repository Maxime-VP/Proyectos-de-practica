#User defined functions

Podemos crear nuestras propias funciones.
return es lo que regresará la función

```python
def calcular_ganancia(ingresos, gastos):
  return ingresos - gastos

ing, gast = 10000, 9001

calcular_ganancia(ing, gast)
```

```python
def calcular_bonus(salary, bonus_rate):
  return salary*bonus_rate

salary = 80000
bonus_rate = .2 # 20%

calcular_bonus(salary, bonus_rate)
```

#Lambda

Funciones anónimas

```python
mul_two = lambda x: x*2
```

```python
mul_two(6)
```

```python
(lambda x, y: x**2+y)(2,2)
```

```python
salary_list = [1000, 2000, 4000,8000]
bonus_rate = .1

total_salary_list = \
 [(lambda salary: salary*(1+bonus_rate))(salary) for salary in salary_list]

print(total_salary_list)
```
