```python
from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .appName("SimpleSparkProgram") \
    .getOrCreate()

print(spark.version)
```

##Example Dataframe

```python
data = [
    ("Max", 22),
    ("Ana", 25),
    ("Carlos", 30),
    ("Luis", 28)
]

columns = ["name", "age"]

df = spark.createDataFrame(data, columns)

df.show()
```

##Example transformation

```python
older_people = df.filter(df.age > 25)

older_people.show()
```
